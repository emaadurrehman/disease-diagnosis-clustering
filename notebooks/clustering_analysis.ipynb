```python
import pandas as pd 
import numpy as np
pd.set_option("display.max_rows", None, "display.max_columns", None)
```

```python
#Loading the Data file
train = pd.read_csv('Data.csv')
train = train.drop(columns='row ID')
print(train.shape)
train.head()
```

```python
#Correlation matrix
cor_matrix = train.corr().abs()
upper_tri = cor_matrix.where(np.triu(np.ones(cor_matrix.shape),k=1).astype(np.bool))
to_drop_train = [column for column in upper_tri.columns if any(upper_tri[column] >= 0.95)]
train = train.drop(to_drop_train, axis=1)

print(train.shape)
```

```python
#Removing features with constant values
temp = []

for c in train.columns:
    if max(train[c])==min(train[c]):
        temp.append(c)
train = train.drop(temp, axis=1)
print(train.shape)
```

```python
#Variance threshold
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0)
selector.fit(train)

train = selector.transform(train)

print(train.shape)
```

```python
#Random state
from sklearn.cluster import KMeans
WSS = []
for k in range(2,500):
    kmeans = KMeans(random_state=k)
    kmeans.fit(train)
    WSS.append(kmeans.inertia_)
print(WSS) 

import matplotlib.pyplot as plt

plt.figure(figsize=(100,10),)
plt.plot(range(2,500),WSS,)
plt.grid(True)
#plt.savefig('random.png')
print(min(WSS))
WSS.index(min(WSS))
```

```python
#Using loop for K-Means Clustering
from sklearn.cluster import KMeans
WSS = []
for k in range(0,10):
    kmeans = KMeans()
    kmeans.fit(train)
    WSS.append(kmeans.inertia_)
    out = pd.read_csv('sample_submission.csv')
    out['cluster'] = kmeans.labels_
    out.to_csv('D:\IDM Challenge 3\Trial '+str(k+45)+'.csv', index = False) 
print(WSS) 

import matplotlib.pyplot as plt

plt.figure(figsize=(100,10),)
plt.plot(range(0,10),WSS,)
plt.grid(True)
#plt.savefig('random.png')
print(min(WSS))
WSS.index(min(WSS))
#8636.1473829201
```

```python
#Agglomerative Clustering Looped
from sklearn.cluster import AgglomerativeClustering


WSS = []
for k in range(0,10):
    agg = AgglomerativeClustering(n_clusters=11,linkage='average',compute_distances=True, affinity = 'manhattan')
    agg.fit(train)
    WSS.append(max(agg.distances_))
    out = pd.read_csv('sample_submission.csv')
    out['cluster'] = agg.labels_
    out.to_csv('D:\IDM Challenge 3\Trial'+str(k+42)+'.csv', index = False) 
print(WSS) 

import matplotlib.pyplot as plt

plt.figure(figsize=(100,10),)
plt.plot(range(0,10),WSS,)
plt.grid(True)
#plt.savefig('random.png')
print(min(WSS))
WSS.index(min(WSS))

print(min(WSS))
print(WSS.index(min(WSS)))
print(max(WSS))
print(WSS.index(max(WSS)))
yes = sorted(range(len(WSS)), key = lambda sub: WSS[sub])[:10]
print(yes)
for l in range(0,9):
    print(WSS[yes[l]])
```

```python
#Elbow Curve
from sklearn.cluster import KMeans
WSS = []
for k in range(2,25):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(train)
    WSS.append(kmeans.inertia_)
print(WSS) 

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(range(2,25),WSS)
plt.grid(True)
#plt.savefig('elbow.png')
from kneed import KneeLocator

kl = KneeLocator(range(2, 25), WSS, curve="convex", direction="decreasing")
kl.elbow
```

```python
#Silhouette Curve
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans

s_avg = []
for k in range(2, 25):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(train)
    sil_sc = silhouette_score(train, kmeans.labels_)
    s_avg.append(sil_sc)
print(s_avg)
print(max(s_avg))
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(range(2,25),s_avg)
plt.grid(True)

plt.show()
plt.savefig('silhouette.png')
print(max(s_avg))
print(s_avg.index((max(s_avg)))+2)
```

```python
#Dendrogram
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

linked = linkage(train, 'ward')
plt.figure(figsize=(100, 5))
dendrogram(linked) 
plt.axhline(y = 19, color = 'b', linestyle = '--')
plt.show()
#plt.savefig('dendrogram.jpeg')
```
